```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("Amazon Sale Report.csv")
df.head()
```

```python
df.columns = df.columns.str.strip()  # removes extra spaces
df.columns = df.columns.str.replace(" ", "_")
df.columns
```

```python
conn = sqlite3.connect("Sales.db")
df.to_sql("Sales", conn, if_exists="replace",index=False)
```

```python
query = "SELECT SUM(Amount) AS Total_Sales FROM Sales"
pd.read_sql(query, conn)
```

```python
query = """
SELECT Category, SUM(Amount) AS total_sales
FROM Sales
GROUP BY Category
ORDER BY total_sales DESC
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT Sales_Channel, SUM(Amount) AS total_sales
FROM Sales
GROUP BY Sales_Channel
ORDER BY total_sales DESC
"""
pd.read_sql(query, conn)
```

```python
pd.read_sql("SELECT * FROM sales LIMIT 1", conn)
```

```python
query = """
SELECT "ship-city", SUM(Amount) AS total_sales
FROM sales
GROUP BY "ship-city"
ORDER BY total_sales DESC
LIMIT 5
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT strftime('%m', Date) AS month,
       SUM(Amount) AS total_sales
FROM sales
GROUP BY month
ORDER BY month
"""
pd.read_sql(query, conn)
```
